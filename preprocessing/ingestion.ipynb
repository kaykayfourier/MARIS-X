```python
import pandas as pd
```

```python
path = "../data/DARTIS_2019.tab"
with open(path, encoding="utf-8") as f:
    skip = next(i for i, line in enumerate(f) if line.strip() == "*/") + 1

df = pd.read_csv(path, sep="\t", skiprows=skip)
print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
img_col = [c for c in df.columns if "IMAGE" in c.upper()][0]
bin_col = [c for c in df.columns if "Binary" in c][0]
print(df[[img_col, bin_col]].head(10).to_string())
```
